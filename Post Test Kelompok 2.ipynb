```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Kelas G_Fixed Call Report.csv')
df.columns = df.columns.str.strip()
df['Month'] = df['Month'].astype(str).str.strip()
months_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'June', 'July', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']
df['Month'] = pd.Categorical(df['Month'], categories=months_order, ordered=True)
print("Data sukses dimuat!")
```

```python
# Kode untuk mengecek rata-rata asli per tahun
print(df.groupby('Year')['Number of Total Ambulance Information'].mean())
```

```python
plt.figure(figsize=(10, 5))
data_a = df.groupby('Year')['Number of Total Ambulance Information'].mean().reset_index()
sns.barplot(data=data_a, x='Year', y='Number of Total Ambulance Information', palette='viridis')
plt.title('Kategori A: Rata-rata Panggilan Info Ambulans per Tahun', fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
data_b = df[df['Year'] == 2023].sort_values('Month')
sns.lineplot(data=data_b, x='Month', y='Total Number of Calls', marker='o', color='coral', linewidth=2.5)
plt.title('Kategori B: Tren Total Panggilan Masuk (Sepanjang Tahun 2023)', fontweight='bold')
plt.xlabel('Bulan (Tahun 2023)')
plt.show()
```

```python
print("--- RINCIAN DATA KATEGORI C (Angka Korelasi Hubungan Dua Variabel) ---")
kolom_korelasi = ['Number of Total Health Information', 'Number of Total Ambulance Information']
print(df[kolom_korelasi].corr())
```

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='Number of Total Health Information', y='Number of Total Ambulance Information', color='teal')
plt.title('Kategori C: Hubungan Info Kesehatan vs Info Ambulans', fontweight='bold')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================================
# 1. LOAD DATASET & PREPARASI DATA (TERMASUK SORTING KATEGORI D)
# =========================================================================
file_path = '/content/Kelas G_Fixed Call Report.csv'
df = pd.read_csv(file_path)

# Membersihkan spasi pada nama kolom dan data teks
df.columns = df.columns.str.strip()
df['Month'] = df['Month'].astype(str).str.strip()

# Mengunci urutan bulan dari Jan sampai Dec agar kronologis saat di-sort
months_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'June', 'July', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']
df['Month'] = pd.Categorical(df['Month'], categories=months_order, ordered=True)

# Urutkan data berdasarkan Tahun dan Bulan dari terlama ke terbaru (Sesuai request)
df_sorted = df.sort_values(['Year', 'Month'], ascending=True).copy()
# Membuat kolom gabungan Bulan-Tahun untuk label sumbu X Kategori D
df_sorted['Month_Year'] = df_sorted['Month'].astype(str) + '-' + df_sorted['Year'].astype(str)

# Preparasi Data A & B
data_a = df.groupby('Year')['Number of Total Ambulance Information'].mean().reset_index()
data_b = df[df['Year'] == 2023].sort_values('Month')

# =========================================================================
# 2. PENGATURAN TATA LETAK DASHBOARD (GRID KUSTOM)
# =========================================================================
sns.set_theme(style="whitegrid")
fig = plt.figure(figsize=(24, 22)) # Ukuran kanvas diperbesar agar muat semua

# Membuat grid kustom: 3 Baris, 4 Kolom
ax_a = plt.subplot2grid((3, 4), (0, 0), colspan=2) # Baris 0, Kolom 0-1
ax_b = plt.subplot2grid((3, 4), (0, 2), colspan=2) # Baris 0, Kolom 2-3
ax_c = plt.subplot2grid((3, 4), (1, 1), colspan=2) # Baris 1, Kolom 1-2 (Tengah)
ax_d = plt.subplot2grid((3, 4), (2, 0), colspan=4) # Baris 2, Kolom 0-3 (Full Lebar)

# =========================================================================
# 3. PLOTTING MASING-MASING KATEGORI
# =========================================================================

# -------------------------------------------------------------------------
# [ATAS KIRI] - KATEGORI A: BAR CHART (Rata-rata Ambulans)
# -------------------------------------------------------------------------
sns.barplot(data=data_a, x='Year', y='Number of Total Ambulance Information', ax=ax_a, palette='viridis')
ax_a.set_title('Kategori A: Rata-rata Panggilan Info Ambulans per Tahun', fontweight='bold', fontsize=13)
ax_a.set_xlabel('Tahun', fontweight='bold')
ax_a.set_ylabel('Rata-rata Jumlah Panggilan', fontweight='bold')

for container in ax_a.containers:
    ax_a.bar_label(container, fmt='%.0f', padding=3, fontweight='bold', fontsize=10)
ax_a.set_ylim(0, data_a['Number of Total Ambulance Information'].max() * 1.15)


# -------------------------------------------------------------------------
# [ATAS KANAN] - KATEGORI B: LINE CHART (Tren 2023)
# -------------------------------------------------------------------------
sns.lineplot(data=data_b, x='Month', y='Total Number of Calls', marker='o', linewidth=2.5, markersize=8, ax=ax_b, color='coral')
ax_b.set_title('Kategori B: Tren Total Panggilan Masuk (Sepanjang 2023)', fontweight='bold', fontsize=13)
ax_b.set_xlabel('Bulan (Tahun 2023)', fontweight='bold')
ax_b.set_ylabel('Total Panggilan Masuk', fontweight='bold')


# -------------------------------------------------------------------------
# [TENGAH] - KATEGORI C: SCATTER + REGRESSION PLOT
# -------------------------------------------------------------------------
sns.regplot(data=df, x='Number of Total Health Information', y='Number of Total Ambulance Information',
            ax=ax_c, color='teal', scatter_kws={'alpha':0.6}, line_kws={'color':'red', 'linewidth':2.5})
ax_c.set_title('Kategori C: Hubungan Info Kesehatan vs Info Ambulans', fontweight='bold', fontsize=13)
ax_c.set_xlabel('Jumlah Info Kesehatan', fontweight='bold')
ax_c.set_ylabel('Jumlah Info Ambulans', fontweight='bold')


# -------------------------------------------------------------------------
# [BAWAH - FULL WIDTH] - KATEGORI D: TIME-SERIES BAR CHART (KODE BARU ANDA)
# -------------------------------------------------------------------------
# Menggunakan data yang sudah disortir kronologis dan menggunakan palette 'magma'
sns.barplot(data=df_sorted, x='Month_Year', y='Total Number of Calls', palette='magma', ax=ax_d)

ax_d.set_title('Kategori D: Total Panggilan Masuk per Bulan (Semua Data)', fontweight='bold', fontsize=14)
ax_d.set_xlabel('Bulan-Tahun', fontweight='bold')
ax_d.set_ylabel('Total Panggilan', fontweight='bold')

# Memutar label sumbu X 90 derajat & mengecilkan font sedikit agar muat 107 bulan
ax_d.tick_params(axis='x', rotation=90, labelsize=9)
ax_d.grid(axis='y', linestyle='--', alpha=0.6)


# =========================================================================
# 4. FINALISASI TAMPILKAN DASHBOARD
# =========================================================================
plt.tight_layout(pad=4.0)
fig.suptitle('Dashboard Visualisasi Data Praktikum (Format Final)', fontsize=20, fontweight='bold', y=1.02)
plt.show()
```
