```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. LOAD DATASET
file_path = '/content/Kelas G_Fixed Call Report.csv'
df = pd.read_csv(file_path)

# Pembersihan spasi gaib pada nama kolom dan isi data bulan
df.columns = df.columns.str.strip()
df['Month'] = df['Month'].astype(str).str.strip()

# Mengatur urutan bulan berdasarkan format teks unik di dataset Anda ('June', 'July', 'Sept')
months_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'June', 'July', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']
df['Month'] = pd.Categorical(df['Month'], categories=months_order, ordered=True)

print("Dataset berhasil dimuat dan siap digunakan!")
```

```python
# Mengatur tema visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Menghitung rata-rata Info Ambulans per Tahun
data_a = df.groupby('Year')['Number of Total Ambulance Information'].mean().reset_index()

# Membuat Bar Chart
sns.barplot(data=data_a, x='Year', y='Number of Total Ambulance Information', palette='viridis')

# Atribut Grafik
plt.title('Kategori A: Rata-rata Panggilan Info Ambulans per Tahun', fontweight='bold', fontsize=14, pad=15)
plt.xlabel('Tahun', fontweight='bold')
plt.ylabel('Rata-rata Jumlah Panggilan', fontweight='bold')

plt.show()
```
