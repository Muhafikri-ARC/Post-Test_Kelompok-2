```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Kelas G_Fixed Call Report.csv')
df.columns = df.columns.str.strip()
df['Month'] = df['Month'].astype(str).str.strip()
months_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'June', 'July', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']
df['Month'] = pd.Categorical(df['Month'], categories=months_order, ordered=True)
print("Data sukses dimuat!")
```

```python
# Kode untuk mengecek rata-rata asli per tahun
print(df.groupby('Year')['Number of Total Ambulance Information'].mean())
```

```python
plt.figure(figsize=(10, 5))
data_a = df.groupby('Year')['Number of Total Ambulance Information'].mean().reset_index()
sns.barplot(data=data_a, x='Year', y='Number of Total Ambulance Information', palette='viridis')
plt.title('Kategori A: Rata-rata Panggilan Info Ambulans per Tahun', fontweight='bold')
plt.show()
```

```python
print("--- RINCIAN DATA KATEGORI B (Total Panggilan Tiap Bulan di Tahun 2023) ---")
# Mengurutkan bulan berdasarkan kalender agar tidak acak
months_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'June', 'July', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']
df['Month'] = pd.Categorical(df['Month'], categories=months_order, ordered=True)

data_b = df[df['Year'] == 2023].sort_values('Month')
print(data_b[['Month', 'Total Number of Calls']].to_string(index=False))
```

```python
plt.figure(figsize=(10, 5))
data_b = df[df['Year'] == 2023].sort_values('Month')
sns.lineplot(data=data_b, x='Month', y='Total Number of Calls', marker='o', color='coral', linewidth=2.5)
plt.title('Kategori B: Tren Total Panggilan Masuk (Sepanjang Tahun 2023)', fontweight='bold')
plt.xlabel('Bulan (Tahun 2023)')
plt.show()
```

```python
print("--- RINCIAN DATA KATEGORI C (Angka Korelasi Hubungan Dua Variabel) ---")
kolom_korelasi = ['Number of Total Health Information', 'Number of Total Ambulance Information']
print(df[kolom_korelasi].corr())
```

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='Number of Total Health Information', y='Number of Total Ambulance Information', color='teal')
plt.title('Kategori C: Hubungan Info Kesehatan vs Info Ambulans', fontweight='bold')
plt.show()
```

```python
import pandas as pd

# 1. Bersihkan spasi gaib di nama kolom agar terhindar dari KeyError
df.columns = df.columns.str.strip()

# 2. Atur format tampilan Pandas agar mencetak angka desimal biasa (menggunakan koma ribuan)
pd.options.display.float_format = '{:,.2f}'.format

print("--- RINCIAN DATA KATEGORI D (Statistik Distribusi Total Panggilan) ---")
# 3. Jalankan describe() untuk melihat statistik deskriptifnya
print(df['Total Number of Calls'].describe())

# Optional: Menambahkan info kemiringan (skewness) untuk memperkuat analisis distribusi
print(f"Skewness (Kemiringan Distribusi): {df['Total Number of Calls'].skew():.2f}")
```
