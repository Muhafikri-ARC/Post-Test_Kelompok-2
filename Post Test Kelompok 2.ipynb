```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Kelas G_Fixed Call Report.csv')


months_order = ['January', 'February', 'March', 'April', 'May', 'June',
                'July', 'August', 'September', 'October', 'November', 'December']
df['Month'] = pd.Categorical(df['Month'], categories=months_order, ordered=True)


# Data A: Rata-rata Info Ambulans per Tahun
data_a = df.groupby('Year')['Number of Total Ambulance Information'].mean().reset_index()

# Data B: Tren Total Panggilan tahun 2023
data_b = df[df['Year'] == 2023].sort_values('Month')


# VISUALISASI GRID 2X2 (Pendekatan Objek)

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Grafik 1: Bar Chart (Kategori A)
sns.barplot(data=data_a, x='Year', y='Number of Total Ambulance Information', ax=axes[0, 0], palette='viridis')
axes[0, 0].set_title('Kategori A: Rata-rata Panggilan Info Ambulans per Tahun', fontweight='bold')

# Grafik 2: Line Chart (Kategori B)
sns.lineplot(data=data_b, x='Month', y='Total Number of Calls', marker='o', ax=axes[0, 1], color='coral')
axes[0, 1].set_title('Kategori B: Tren Total Panggilan (Tahun 2023)', fontweight='bold')
axes[0, 1].tick_params(axis='x', rotation=45)

```
