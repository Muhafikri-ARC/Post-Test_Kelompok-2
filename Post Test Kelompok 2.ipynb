```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. LOAD DATASET
file_path = '/content/Kelas G_Fixed Call Report.csv'
df = pd.read_csv(file_path)

# Pembersihan spasi gaib pada nama kolom dan isi data bulan
df.columns = df.columns.str.strip()
df['Month'] = df['Month'].astype(str).str.strip()

# Mengatur urutan bulan berdasarkan format teks unik di dataset Anda ('June', 'July', 'Sept')
months_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'June', 'July', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']
df['Month'] = pd.Categorical(df['Month'], categories=months_order, ordered=True)

print("Dataset berhasil dimuat dan siap digunakan!")
```

```python
# Mengatur tema visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Menghitung rata-rata Info Ambulans per Tahun
data_a = df.groupby('Year')['Number of Total Ambulance Information'].mean().reset_index()

# Membuat Bar Chart
sns.barplot(data=data_a, x='Year', y='Number of Total Ambulance Information', palette='viridis')

# Atribut Grafik
plt.title('Kategori A: Rata-rata Panggilan Info Ambulans per Tahun', fontweight='bold', fontsize=14, pad=15)
plt.xlabel('Tahun', fontweight='bold')
plt.ylabel('Rata-rata Jumlah Panggilan', fontweight='bold')

plt.show()
```

```python
# Mengatur tema visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# Menyaring tren data khusus untuk tahun 2024 dan diurutkan berdasarkan bulan
data_b = df[df['Year'] == 2024].sort_values('Month')

# Membuat Line Chart
sns.lineplot(data=data_b, x='Month', y='Total Number of Calls', marker='o', linewidth=2.5, markersize=8, color='coral')

# Atribut Grafik
plt.title('Kategori B: Tren Total Panggilan Masuk (Tahun 2024)', fontweight='bold', fontsize=14, pad=15)
plt.xlabel('Bulan', fontweight='bold')
plt.ylabel('Total Panggilan', fontweight='bold')
plt.xticks(rotation=30) # Memiringkan tulisan bulan agar tidak bertumpuk

plt.show()
```

```python
# Mengatur tema visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Membuat Scatter Plot korelasi antara Info Kesehatan dan Info Ambulans
sns.scatterplot(data=df, x='Number of Total Health Information', y='Number of Total Ambulance Information', color='teal', alpha=0.7, s=60)

# Atribut Grafik
plt.title('Kategori C: Hubungan Info Kesehatan vs Info Ambulans', fontweight='bold', fontsize=14, pad=15)
plt.xlabel('Jumlah Info Kesehatan', fontweight='bold')
plt.ylabel('Jumlah Info Ambulans', fontweight='bold')

plt.show()
```

```python
# Mengatur tema visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Membuat Histogram untuk melihat sebaran/distribusi total panggilan bulanan
sns.histplot(data=df, x='Total Number of Calls', kde=True, bins=15, color='purple')

# Atribut Grafik
plt.title('Kategori D: Distribusi Frekuensi Total Panggilan Masuk Bulanan', fontweight='bold', fontsize=14, pad=15)
plt.xlabel('Total Panggilan Bulanan', fontweight='bold')
plt.ylabel('Frekuensi Muncul', fontweight='bold')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 1. LOAD DATASET (Path yang benar)
# ==========================================
# File yang diunggah ke sidebar Colab otomatis ada di /content/
df = pd.read_csv('/content/Kelas G_Fixed Call Report.csv')

# Pre-processing: Mengatur urutan bulan
months_order = ['January', 'February', 'March', 'April', 'May', 'June',
                'July', 'August', 'September', 'October', 'November', 'December']
df['Month'] = pd.Categorical(df['Month'], categories=months_order, ordered=True)

# ==========================================
# 2. PERSIAPAN DATA UNTUK VISUALISASI
# ==========================================
# Data A: Rata-rata Info Ambulans per Tahun
data_a = df.groupby('Year')['Number of Total Ambulance Information'].mean().reset_index()

# Data B: Tren Total Panggilan tahun 2023
data_b = df[df['Year'] == 2023].sort_values('Month')

# ==========================================
# 3. VISUALISASI GRID 2X2 (Pendekatan Objek)
# ==========================================
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Grafik 1: Bar Chart (Kategori A)
sns.barplot(data=data_a, x='Year', y='Number of Total Ambulance Information', ax=axes[0, 0], palette='viridis')
axes[0, 0].set_title('Kategori A: Rata-rata Panggilan Info Ambulans per Tahun', fontweight='bold')

# Grafik 2: Line Chart (Kategori B)
sns.lineplot(data=data_b, x='Month', y='Total Number of Calls', marker='o', ax=axes[0, 1], color='coral')
axes[0, 1].set_title('Kategori B: Tren Total Panggilan (Tahun 2023)', fontweight='bold')
axes[0, 1].tick_params(axis='x', rotation=45)

# Grafik 3: Scatter Plot (Kategori C)
sns.scatterplot(data=df, x='Number of Total Health Information', y='Number of Total Ambulance Information', ax=axes[1, 0], color='teal')
axes[1, 0].set_title('Kategori C: Korelasi Info Kesehatan vs Ambulans', fontweight='bold')

# Grafik 4: Histogram (Kategori D)
sns.histplot(data=df, x='Total Number of Calls', kde=True, bins=15, ax=axes[1, 1], color='purple')
axes[1, 1].set_title('Kategori D: Distribusi Total Panggilan Masuk', fontweight='bold')

# Finalisasi
plt.tight_layout(pad=3.0)
fig.suptitle('Dashboard Laporan Panggilan - Kelompok 2', fontsize=18, fontweight='bold', y=1.02)
plt.show()
```
