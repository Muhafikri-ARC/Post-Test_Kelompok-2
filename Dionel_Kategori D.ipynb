```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/Kelas G_Fixed Call Report.csv', sep=',',decimal=';')
```

```python
df
```

```python
import pandas as pd

# 1. Bersihkan spasi gaib di nama kolom agar terhindar dari KeyError
df.columns = df.columns.str.strip()

# 2. Atur format tampilan Pandas agar mencetak angka desimal biasa (menggunakan koma ribuan)
pd.options.display.float_format = '{:,.2f}'.format

print("--- RINCIAN DATA KATEGORI D (Statistik Distribusi Total Panggilan) ---")
# 3. Jalankan describe() untuk melihat statistik deskriptifnya
print(df['Total Number of Calls'].describe())

# Optional: Menambahkan info kemiringan (skewness) untuk memperkuat analisis distribusi
print(f"Skewness (Kemiringan Distribusi): {df['Total Number of Calls'].skew():.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Membuat kolom gabungan Bulan-Tahun untuk label sumbu X
df['Month_Year'] = df['Month'].astype(str) + '-' + df['Year'].astype(str)

# Mengurutkan data berdasarkan Tahun dan Bulan agar urutan di grafik logis
# (Berdasarkan urutan data asli dari terbaru ke terlama atau sebaliknya)
df_sorted = df.sort_values(['Year', 'Month'], ascending=True)

plt.figure(figsize=(25, 8))
sns.barplot(data=df_sorted, x='Month_Year', y='Total Number of Calls', palette='magma')

plt.title('Total Panggilan Masuk per Bulan (Semua Data)', fontweight='bold', fontsize=16)
plt.xlabel('Bulan-Tahun', fontweight='bold')
plt.ylabel('Total Panggilan', fontweight='bold')
plt.xticks(rotation=90) # Memutar label agar tidak tumpang tindih
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
